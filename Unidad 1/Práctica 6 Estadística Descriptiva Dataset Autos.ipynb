```python
import pandas as pd

autos = pd.read_csv('autos2.csv', encoding='ISO-8859-1')
display(autos.head())
autos.info()
```

```python
import matplotlib.pyplot as plt

print("Estadísticas del Precio:")
display(autos['price'].describe())

# Histograma de Precios
media_precio = autos['price'].mean()
plt.figure(figsize=(10, 6))
plt.hist(autos['price'].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.axvline(media_precio, color='red', linestyle='dashed', linewidth=2, label=f'Media Precio=${media_precio:.2f}')
plt.title("Histograma de Precios de Autos")
plt.xlabel("Precio")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Gráfica de barras por Tipo de Combustible
conteo_combustible = autos['fuel-type'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(conteo_combustible.index, conteo_combustible.values, color='lightgreen', edgecolor='black', alpha=0.7)
plt.title("Distribución por Tipo de Combustible")
plt.xlabel("Combustible")
plt.ylabel("Cantidad de Autos")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Boxplot: Precio vs Tipo de Combustible
plt.figure(figsize=(10, 6))
sns.boxplot(x='fuel-type', y='price', data=autos, palette='pastel', hue='fuel-type', legend=False)
plt.title("Distribución de Precio por Tipo de Combustible", fontsize=14, fontweight='bold')
plt.xlabel("Tipo de Combustible", fontsize=12)
plt.ylabel("Precio", fontsize=12)
plt.show()
```

## Análisis Estadístico Descriptivo: Autos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Estadísticas Descriptivas
print("--- Resumen Estadístico de Precios ---")
display(autos['price'].describe())

# 2. Histograma con Media
plt.figure(figsize=(10, 5))
media_p = autos['price'].mean()
plt.hist(autos['price'].dropna(), bins=15, color='#86bf91', edgecolor='black', rwidth=0.9)
plt.axvline(media_p, color='red', linestyle='--', linewidth=2, label=f'Media: ${media_p:.2f}')
plt.title('Distribución de Precios de Automóviles')
plt.xlabel('Precio (USD)')
plt.ylabel('Número de Autos')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# 3. Gráfica de Barras: Tipos de Combustible
plt.figure(figsize=(8, 5))
autos['fuel-type'].value_counts().plot(kind='bar', color=['orange', 'blue'], edgecolor='black')
plt.title('Cantidad de Autos por Tipo de Combustible')
plt.xlabel('Tipo de Combustible')
plt.ylabel('Frecuencia')
plt.xticks(rotation=0)
plt.show()
```

```python
# 4. Boxplot Comparativo: Precio por Combustible
plt.figure(figsize=(10, 6))
sns.boxplot(x='fuel-type', y='price', data=autos, hue='fuel-type', palette='Set2', legend=False)
plt.title('Comparativa de Precios: Gasolina vs Diesel', fontsize=14)
plt.xlabel('Combustible')
plt.ylabel('Precio')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configuración de estilo
sns.set_style("whitegrid")

# 1. Estadísticas detalladas de la variable cuantitativa (Precio)
print(f"Valor Máximo: {autos.price.max()}")
print(f"Valor Mínimo: {autos.price.min()}")
print("--- Resumen completo ---")
display(autos.price.describe())

# 2. Histograma con bins específicos y línea de media
plt.figure(figsize=(10, 6))
media = autos.price.mean()
# Definimos rangos dinámicos para el histograma basados en el precio
plt.hist(autos.price.dropna(), bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(media, color='red', linestyle='dashed', linewidth=2, label=f'Media = ${media:.2f}')
plt.title("Histograma de Precios (Requisito: Distribución y Media)", fontsize=14)
plt.xlabel("Precio")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()

# 3. Boxplots Comparativos (Requisito: Análisis de varianza visual)
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(x="fuel-type", y="price", data=autos, palette="pastel", ax=ax[0], hue="fuel-type", legend=False)
ax[0].set_title("Precio por Tipo de Combustible", fontweight="bold")

sns.boxplot(x="body-style", y="price", data=autos, palette="flare", ax=ax[1], hue="body-style", legend=False)
ax[1].set_title("Precio por Estilo de Carrocería", fontweight="bold")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**Conclusión del Análisis:**
Este análisis exploratorio de los datos de automóviles nos ha permitido identificar varias características clave:

*   **Distribución de Precios:** La mayoría de los vehículos se concentran en rangos de precios más bajos, con una cola hacia precios más altos, indicando la presencia de vehículos de lujo o de alto rendimiento. La media de precios se sitúa alrededor de los $13,207.

*   **Impacto del Tipo de Combustible en el Precio:** Los boxplots revelan que los autos de **Diesel** tienden a tener un rango de precio base más alto en comparación con los de **Gasolina**. Sin embargo, es interesante notar que los precios máximos (identificados como *outliers*) se observan en los vehículos de gasolina, sugiriendo que los autos de lujo o deportivos suelen usar este tipo de combustible. Esto implica que, aunque el diésel puede tener un costo inicial más elevado en promedio, la gasolina domina el segmento de precios extremos.

*   **Influencia del Estilo de Carrocería en el Precio:** El estilo de carrocería también juega un papel significativo en la determinación del precio. Por ejemplo, los 'hardtop' y 'convertible' muestran distribuciones de precios más altas, lo cual es esperable dada su naturaleza a menudo más premium o especializada, mientras que otros estilos como 'sedan' o 'hatchback' cubren un rango más amplio de precios, incluyendo opciones más económicas.

Este análisis proporciona una base sólida para entender las dinámicas de precios en el mercado automotriz, destacando la importancia de variables como el tipo de combustible y el estilo de carrocería en la valoración de los vehículos.
