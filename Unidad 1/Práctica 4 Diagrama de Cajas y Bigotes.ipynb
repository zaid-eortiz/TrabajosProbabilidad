```python
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
fabrica1 = []
for i in range(100):
  numero_aleatorio = round(random.uniform(50, 100), 2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio2 = round(random.uniform(70, 100), 2)
  fabrica2.append(numero_aleatorio2)
print(fabrica2)
```

##**Diagrama de Cajas y Bigotes**

1.   Las lineas se usan para indicar la variabilidad fuera de los cuartiles superior e inferior.
2.   Los valores atípicos se representan a veces como puntos individuales que estan en linea con los bigotes.
3.   **Si los datos son simetricos, la mediana estara en el centro de la caja.**
4.   Si los datos estan **sesgados**, la mediana estara mas cerca de la parte superior o inferior de la caja.



```python
plt.boxplot(fabrica1)
plt.ylabel('Edades')
plt.title('Produccion Fabrica 1')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(nrows =1, ncols =2, figsize=(9, 4))
bplot = ax1.boxplot(fabrica1,
                    vert = True, #vertical box alignment
                    patch_artist=True) #fill with color
ax1.set_title('Produccion Fabrica 1')
bplot = ax2.boxplot(fabrica2,
                    vert = True, #vertical box alignment
                    patch_artist=True) #fill with color
ax2.set_title('Produccion Fabrica 2')
plt.show()
```
