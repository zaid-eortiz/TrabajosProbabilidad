```python
import matplotlib.pyplot as plt
import numpy as np

# Datos del Ejercicio 3
x = [0, 1, 2, 3, 4, 5]
p = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10]

# 1. Verificar que las probabilidades sumen 1
suma_prob = sum(p)
print(f"1. Suma de probabilidades: {suma_prob:.2f}")

# 2. Distribución de probabilidad
print("\n2. Distribución de Probabilidad:")
for xi, pi in zip(x, p):
    print(f"Unidades = {xi} | P(X=x) = {pi}")

# 3. Valor esperado, Varianza y Desviación Estándar
esperanza = sum(xi * pi for xi, pi in zip(x, p))
varianza = sum(((xi - esperanza)**2) * pi for xi, pi in zip(x, p))
desviacion_estandar = varianza ** 0.5

print(f"\n3. Estadísticas:")
print(f"Valor Esperado E(X): {esperanza:.2f} unidades")
print(f"Varianza: {varianza:.2f}")
print(f"Desviación Estándar: {desviacion_estandar:.2f}")

# 4. Función de distribución acumulada F(x)
F = np.cumsum(p)
print("\n4. Función de Distribución Acumulada F(x):")
for xi, Fi in zip(x, F):
    print(f"F({xi}) = {Fi:.2f}")

# Generación de Gráficas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfica de Distribución de Probabilidad
ax1.stem(x, p, basefmt=" ")
ax1.set_title("Distribución de Probabilidad P(x)")
ax1.set_xlabel("Unidades vendidas por semana (x)")
ax1.set_ylabel("Probabilidad")
ax1.set_xticks(x)
ax1.grid(True, linestyle='--', alpha=0.7)

# Gráfica de Distribución Acumulada
ax2.step(x, F, where='post', marker='o', color='orange')
ax2.set_title("Distribución Acumulada F(x)")
ax2.set_xlabel("Unidades vendidas por semana (x)")
ax2.set_ylabel("Probabilidad Acumulada")
ax2.set_xticks(x)
ax2.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
