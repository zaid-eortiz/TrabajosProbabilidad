```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. DATOS Y SOLUCIÓN DEL PROBLEMA

media = 70        # Media en kg
desv_std = 4      # Desviación estándar en kg
num_datos = 500   # Número de estudiantes
limite_inf = 68
limite_sup = 72

# Cálculo de probabilidad y número de estudiantes
probabilidad = norm.cdf(limite_sup, media, desv_std) - norm.cdf(limite_inf, media, desv_std)
estudiantes_rango = probabilidad * num_datos

print("--- RESPUESTAS DEL PROBLEMA 1 ---")
print(f"La probabilidad de estudiantes que pesan entre 68 y 72 kg es: {probabilidad:.4f}")
print(f"La cantidad de estudiantes en ese rango es: {round(estudiantes_rango)}")
print("---------------------------------\n")


# 2. GENERACIÓN DE DATOS

# Generamos los datos aleatorios
datos = np.random.normal(media, desv_std, num_datos)

# Crear un rango de valores para crear una densidad
x = np.linspace(media - 3 * desv_std, media + 3 * desv_std, num_datos)

# Calcular la PDF de la distribución normal usando la fórmula de la clase
pdf = 1 / (desv_std * np.sqrt(2 * np.pi)) * np.exp(-(x - media)**2 / (2 * desv_std **2))



# 3. GRÁFICA 1: SOLO EL HISTOGRAMA

plt.figure(figsize=(6, 4))
plt.hist(datos, bins=15, alpha=0.4, color='b') # Histograma de los datos
plt.title('Distribución Normal ')
plt.xlabel('Peso en kg de los estudiantes')
plt.grid(True)
print("Mostrando Gráfica 1: Histograma...")
plt.show()



# 4. GRÁFICA 2: SOLO LA FUNCIÓN DE DENSIDAD (PDF)

plt.figure(figsize=(6, 4))
plt.plot(x, pdf, label='Normal:(media:{}; ds:{})'.format(np.round(media, 2), np.round(desv_std, 2)))
plt.title('Función de Densidad de Probabilidad - Distribución Normal')
plt.xlabel('Pesos en Kg. de los estudiantes')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True)
print("Mostrando Gráfica 2: Función de Densidad de Probabilidad...")
plt.show()



# 5. GRÁFICA 3: HISTOGRAMA Y PDF UNIDOS

plt.figure(figsize=(6, 4))
# Aquí se usa density=True para que las escalas coincidan
plt.hist(datos, bins=15, alpha=0.4, density=True, color='b')
plt.plot(x, pdf, 'k', linewidth=2)

plt.title('Distribución Normal ')
plt.xlabel('Peso en kg de los estudiantes')
plt.grid(True)
print("Mostrando Gráfica 3: Histograma y PDF unidos...")
plt.show()
```
