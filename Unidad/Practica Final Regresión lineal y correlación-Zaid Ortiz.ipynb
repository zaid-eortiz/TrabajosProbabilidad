```python
def analizar_grupo(archivo):
    print(f"\n{'='*50}")
    print(f"ANÁLISIS DEL DATASET: {archivo}")
    print(f"{'='*50}")

    df = pd.read_csv(archivo)

    # 1. Carga y Exploración
    print("\n▶ Primeras 5 filas del dataset:")
    display(df.head())

    print("\n▶ Descripción estadística:")
    display(df.describe())

    # Histograma
    plt.figure(figsize=(6, 4))
    sns.histplot(df['Calificacion'], bins=10, kde=True, color='skyblue')
    plt.title(f'Histograma de Calificaciones - {archivo}')
    plt.xlabel('Calificación')
    plt.ylabel('Frecuencia')
    plt.show()

    # 2 y 3. Regresión Lineal y Correlación
    X = df[['HorasEstudio']]
    y = df['Calificacion']

    modelo = LinearRegression()
    modelo.fit(X, y)
    correlacion, p_valor = pearsonr(df['HorasEstudio'], df['Calificacion'])
    r2 = modelo.score(X, y)

    print("\n▶ Resultados del Modelo y Correlación:")
    print(f"Ecuación: Calificación = {modelo.intercept_:.4f} + ({modelo.coef_[0]:.4f} * Horas de Estudio)")
    print(f"Coeficiente de Pearson (r): {correlacion:.4f}")
    print(f"Valor p (p-value): {p_valor:.4e}")
    print(f"Coeficiente de determinación (R²): {r2:.4f}")

    # 4. Visualización de Resultados
    plt.figure(figsize=(6, 4))
    sns.scatterplot(x=df['HorasEstudio'], y=df['Calificacion'], color='blue', label='Datos reales')
    plt.plot(df['HorasEstudio'], modelo.predict(X), color='red', label='Línea de regresión')
    plt.title(f'Regresión Lineal: Horas vs Calificación - {archivo}')
    plt.xlabel('Horas de Estudio')
    plt.ylabel('Calificación Final')
    plt.legend()
    plt.show()
```

```python
analizar_grupo('grupo1.csv')
```
