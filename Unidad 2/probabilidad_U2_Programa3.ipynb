```python
import math as mt
from itertools import permutations

# Datos del problema
asistentes = [f"Asistente_{i}" for i in range(1, 11)]  # 10 asistentes
n = len(asistentes)
r = 4  # 4 preguntas

# Cálculo matemático: Pr = n! / (n-r)!
Pr = mt.factorial(n) / mt.factorial(n - r)

print("Existen", int(Pr), "posibles formas de elegir a los asistentes para calificar las 4 preguntas.")
print("Aquí las primeras 100 permutaciones de muestra:")

# Generar y mostrar permutaciones
permutaciones_asistentes = list(permutations(asistentes, r))

# Se imprimen las primeras 100 como muestra para no saturar la consola con las 5,040
for permutacion in permutaciones_asistentes[:100]:
    print(permutacion)
```
